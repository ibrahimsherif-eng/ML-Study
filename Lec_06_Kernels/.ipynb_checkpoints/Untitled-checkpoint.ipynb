```python
sns.set_context("notebook")
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex",
    style="time",
    color="red"
               )

plt.title("first seaborn plot ")
plt.show
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    hue="sex",
    multiple="stack"
)
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    hue="sex"
)

plt.show()
```

```python
sns.ecdfplot(
    data=df,
    x="tip"
)

plt.show()
```
